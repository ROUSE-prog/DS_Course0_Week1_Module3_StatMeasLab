# Widget Sales: Statistical Measures Analysis

## Business Context

Imagine you are a junior analyst at a company that sells widgets across many industries and markets. Your manager wants a concise summary of the past year's performance.

This analysis answers three business questions:

1. **What has been the typical sales volume across markets?**
2. **How variable have sales been across the different markets?**
3. **How much has the company characteristically spent on TV, radio, and newspaper advertising?**

The dataset contains 200 market-level observations. Sales are measured in **thousands of widgets**, while advertising expenditures are measured in **thousands of dollars**.

## Analysis Objectives

We will:

- load the CSV data as a list of dictionaries using `csv.DictReader`
- extract sales and advertising variables into numeric lists
- summarize the range of sales across markets
- visualize the sales distribution
- choose and calculate an appropriate measure of central tendency
- choose and calculate an appropriate measure of dispersion
- compare typical advertising spending and variability across media

```python
# CodeGrade step0
# Run this cell without changes
import csv
import numpy as np
import scipy.stats as stats
```

## Step 1 — Load the Data

### Task

Use `csv.DictReader` to load `Advertising.csv` into a **list of dictionaries** named `data`.

`DictReader` is useful here because each CSV row is represented by a dictionary keyed by the column names, making later extraction more readable.

```python
# CodeGrade step1
# Load the CSV into a list of dictionaries
with open("Advertising.csv") as f:
    data = list(csv.DictReader(f))

# Inspect the first record
data[0]
```

### Analyst Note

At this stage the values are still strings because CSV files do not preserve Python numeric types. We will convert the columns we need to `float` values in the next step.

## Step 2 — Extract the Variables

### Task

Extract four numeric lists:

- `sales`: widgets sold, in thousands
- `tv`: TV advertising spend, in thousands of dollars
- `radio`: radio advertising spend, in thousands of dollars
- `newspaper`: newspaper advertising spend, in thousands of dollars

Converting the values to floats now allows us to calculate statistics later.

```python
# CodeGrade step2
sales = [float(row["sales"]) for row in data]
tv = [float(row["TV"]) for row in data]
radio = [float(row["radio"]) for row in data]
newspaper = [float(row["newspaper"]) for row in data]

# Confirm that all four variables contain one value per market
len(sales), len(tv), len(radio), len(newspaper)
```

### Analyst Note

Each list should contain **200 values**, one for every market in the dataset. Keeping each measure in its own list makes it easy to use Python, NumPy, and SciPy statistical functions.

## Step 3 — Summarize Sales Across Markets

### Task

Provide an initial sales summary by finding:

- the number of markets represented
- the minimum sales observed in any market
- the maximum sales observed in any market

Use Python's built-in `len()`, `min()`, and `max()` functions.

```python
# CodeGrade step3
num_markets = len(sales)
min_sales = min(sales)
max_sales = max(sales)
```

```python
# Run this cell without changes
print(f"""
This dataset contains records for {num_markets} markets

The fewest sales for any market was {min_sales} thousand widgets

The most sales for any market was {max_sales} thousand widgets
""")
```

### Initial Finding

The company operated across **200 markets**. Market-level sales ranged from **1.6 thousand widgets** to **27.0 thousand widgets**.

The range is useful as a first look, but it does not tell us what a *typical* market looks like or how concentrated the middle of the distribution is. For that, we need measures of central tendency and dispersion.

### Visualize the Sales Distribution

A histogram helps us inspect the shape of the sales distribution before choosing summary statistics. This is important because the best measure of a 'typical' value depends on the shape and potential skew of the data.

```python
# Run this cell without changes
import matplotlib.pyplot as plt
%matplotlib inline

fig, ax = plt.subplots(figsize=(10, 5))

ax.hist(sales, bins=15)

ax.set_xlabel("Sales (thousands of widgets)")
ax.set_ylabel("Count")

ax.set_title("Distribution of Sales across Markets");
```

## Step 4 — Typical Sales Volume

### Business Question

**What has been the typical sales volume across markets?**

Based on the histogram, use an appropriate measure of central tendency. We will use the **median**, which describes the midpoint of the observed market sales and is less sensitive to unusually high or low values than the mean.

```python
# CodeGrade step4
# Assign measure_central_tendency to the median of the sales data
measure_central_tendency = np.median(sales)

print(f"Typical sales volume: {measure_central_tendency} thousand widgets")
```

### Interpretation

The median sales volume is **12.9 thousand widgets**. In practical terms, half of the markets sold fewer than 12.9 thousand widgets and half sold more.

This gives management a useful benchmark for what a characteristic market looked like during the year.

## Step 5 — Variability in Sales

### Business Question

**How variable have sales been across markets?**

Because we summarized the center with the median, we will pair it with the **interquartile range (IQR)**. The IQR measures the width of the middle 50% of the observations and is resistant to extreme values.

```python
# CodeGrade step5
# Assign measure_dispersion to the IQR of sales
measure_dispersion = stats.iqr(sales)

print(f"Sales IQR: {measure_dispersion} thousand widgets")
```

### Interpretation

The sales IQR is approximately **7.025 thousand widgets**. This means the middle 50% of markets span a sales interval about 7.0 thousand widgets wide.

Together, the **median (12.9)** and **IQR (7.025)** provide a robust description of the typical sales level and the amount of market-to-market variation.

## Step 6 — Typical Advertising Expenditure

### Business Question

**How much has the company characteristically been spending on different advertising media across markets?**

Calculate the **median expenditure** for TV, radio, and newspaper advertising. Because all advertising variables are measured in thousands of dollars, the medians can be compared directly.

```python
# CodeGrade step6
# Calculate median advertising expenditure
median_tv_expenditure = np.median(tv)
median_radio_expenditure = np.median(radio)
median_newspaper_expenditure = np.median(newspaper)

median_tv_expenditure, median_radio_expenditure, median_newspaper_expenditure
```

### Interpretation

The characteristic advertising spend per market was:

- **TV:** $149.75 thousand
- **Radio:** $22.90 thousand
- **Newspaper:** $25.75 thousand

TV advertising received by far the largest typical allocation. Median newspaper spending was slightly higher than median radio spending.

## Step 7 — Variability in Advertising Expenditure

### Task

Calculate the **IQR** for TV, radio, and newspaper expenditure to describe how much advertising budgets varied across the middle 50% of markets.

```python
# CodeGrade step7
iqr_tv_expenditure = stats.iqr(tv)
iqr_radio_expenditure = stats.iqr(radio)
iqr_newspaper_expenditure = stats.iqr(newspaper)

iqr_tv_expenditure, iqr_radio_expenditure, iqr_newspaper_expenditure
```

### Interpretation

The advertising-spend IQRs are approximately:

- **TV:** $144.45 thousand
- **Radio:** $26.55 thousand
- **Newspaper:** $32.35 thousand

TV spending varied the most in absolute dollar terms across markets. Newspaper spending was also somewhat more variable than radio spending among the middle 50% of markets.

## Business Summary

The analysis provides direct answers to the manager's three questions:

**Typical sales volume:** A characteristic market sold **12.9 thousand widgets**, based on the median.

**Sales variability:** The middle 50% of markets covered an IQR of about **7.025 thousand widgets**, showing meaningful variation in sales performance from market to market.

**Typical advertising spend:** Median market-level expenditures were **$149.75K on TV**, **$22.90K on radio**, and **$25.75K on newspaper**. TV was the dominant advertising channel by typical spend.

### Key Takeaway

The company operated in a wide range of market conditions. A median-and-IQR summary gives management a more representative picture of a typical market than relying only on minimums, maximums, or averages. This analysis establishes a descriptive baseline that could later be extended to examine whether higher advertising spending is associated with higher widget sales.
